## 1. A practical evidence model
A reproducible scientific workflow is supported by connected evidence: documentation, dependencies, environments, tests, provenance, portable execution information, and licensing. Presence is useful evidence, but presence alone does not establish quality or scientific correctness.

```python
evidence = {
    'Documentation': {'examples': ['README.md'], 'supports': 'How to understand and run the project', 'does_not_prove': 'Instructions are complete or correct'},
    'Dependencies': {'examples': ['pyproject.toml', 'requirements.txt'], 'supports': 'Required software is declared', 'does_not_prove': 'Exact versions can be reconstructed'},
    'Environment': {'examples': ['environment.yml', 'lock file', 'spack.yaml'], 'supports': 'Execution environment is described', 'does_not_prove': 'Equivalent behavior on every platform'},
    'Testing': {'examples': ['tests/', 'pytest.ini'], 'supports': 'Automated checks exist', 'does_not_prove': 'Scientific conclusions are valid'},
    'Provenance': {'examples': ['params.yaml', 'provenance.json'], 'supports': 'Results can be connected to configurations', 'does_not_prove': 'Data and models remain accessible'},
    'License': {'examples': ['LICENSE'], 'supports': 'Reuse terms are stated', 'does_not_prove': 'All external artifacts have compatible terms'}
}
for category, item in evidence.items():
    print(f"{category}: {', '.join(item['examples'])}")
    print(f"  Supports: {item['supports']}")
    print(f"  Does not prove: {item['does_not_prove']}")
```

## 2. Audit a repository snapshot
Edit `observed_files` to represent files you observe in a scientific software repository. The exercise intentionally uses a simple presence check so you can identify its limitations.

```python
observed_files = {'README.md', 'pyproject.toml', 'tests', 'LICENSE'}
checks = {
    'Documentation': {'README.md', 'README.rst'},
    'Dependencies': {'pyproject.toml', 'requirements.txt'},
    'Environment': {'environment.yml', 'environment.yaml', 'spack.yaml'},
    'Testing': {'tests', 'test', 'pytest.ini'},
    'Provenance': {'params.yaml', 'provenance.json', 'dvc.yaml'},
    'License': {'LICENSE', 'LICENSE.md', 'COPYING'},
}
results = {}
for category, candidates in checks.items():
    found = sorted(observed_files & candidates)
    results[category] = found
    print(f"{category:15} {'FOUND' if found else 'NOT OBSERVED':12} {', '.join(found) or '-'}")
```

## 3. Interpret, do not overclaim
For every `NOT OBSERVED` result, ask whether the artifact is actually applicable and whether equivalent evidence exists elsewhere. For every `FOUND` result, inspect its quality. A README can exist without run instructions; a test directory can exist without meaningful scientific validation.

### Takeaway
Repository assessment is a **reproducibility-readiness** activity. It helps researchers find evidence and gaps; it is not proof that a scientific result has been reproduced. Continue to Tutorial 2 to make dependency and environment evidence stronger.
