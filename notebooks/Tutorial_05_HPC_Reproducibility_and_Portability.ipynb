## 1. Why HPC context matters
Scientific results on HPC systems may depend on compiler versions, MPI libraries, accelerator stacks, modules, process/thread placement, numerical libraries, and architecture. Scheduler directives also matter operationally, but not every scheduler setting is scientifically essential.

```python
hpc_record = {
    'system': 'example-cluster',
    'architecture': 'x86_64',
    'compiler': 'gcc/13.2',
    'mpi': 'openmpi/4.1',
    'accelerator': 'N/A',
    'modules': ['gcc/13.2','openmpi/4.1'],
    'nodes': 2,
    'tasks_per_node': 8,
    'threads_per_task': 1,
    'scheduler': 'Slurm',
    'walltime': '00:30:00',
    'random_seed': 42,
}
for k,v in hpc_record.items(): print(f'{k:20}: {v}')
```

## 2. Classify portability requirements
The classification below is a starting point, not a universal rule. Whether a setting is scientifically essential depends on the application, numerical sensitivity, scaling regime, and research question.

```python
classification = {
    'compiler': 'record and test portability',
    'mpi': 'record; may affect behavior/performance',
    'nodes': 'method/performance dependent',
    'tasks_per_node': 'method/performance dependent',
    'scheduler': 'usually site-specific',
    'walltime': 'usually site-specific operational setting',
    'random_seed': 'scientifically relevant when stochastic',
}
for setting, meaning in classification.items(): print(f'{setting:18} -> {meaning}')
```

## 3. Translate a job concept across systems
A portable workflow should separate the scientific command from scheduler syntax where possible. The same executable may be launched through Slurm, PBS, LSF, or another scheduler with different directives.

```python
scientific_command = 'python train.py --seed 42 --config config.yaml'
slurm_example = f'''#SBATCH --nodes=2
#SBATCH --ntasks-per-node=8
#SBATCH --time=00:30:00
srun {scientific_command}'''
print(slurm_example)
```
