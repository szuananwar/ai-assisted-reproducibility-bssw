## Learning objectives
Ground AI assistance in verified evidence, separate evidence generation from explanation/prioritization, detect unsupported AI claims, and define points that require human scientific review.

**Estimated time:** 35–50 minutes. No external AI service is required.

## 1. Start from verified findings
AI should not invent the evidence on which a reproducibility assessment depends. A safer pattern is: deterministic or human-verified evidence first, optional AI explanation/prioritization second, human review before action.

```python
verified_findings = [
    {'id':'F1','finding':'No lock file observed','evidence':['pyproject.toml'],'recommendation':'Record a resolved environment for published results.'},
    {'id':'F2','finding':'Automated tests observed','evidence':['tests/','pytest.ini'],'recommendation':'Inspect whether tests include scientific validation.'},
    {'id':'F3','finding':'No provenance record observed','evidence':[],'recommendation':'Record code, data/model, parameters, environment, and result provenance.'},
]
for f in verified_findings: print(f"{f['id']}: {f['finding']} | evidence={f['evidence'] or ['none observed']}")
```

## 2. Evaluate candidate AI statements
Imagine an AI assistant produced the statements below. Classify them against the verified evidence. The point is not whether a statement sounds plausible; it is whether the available evidence supports it.

```python
candidate_ai = [
    {'text':'The project should record a resolved environment because no lock file was observed.','supported_by':['F1']},
    {'text':'The software is scientifically correct because pytest is configured.','supported_by':[]},
    {'text':'Provenance documentation is a high-priority improvement.','supported_by':['F3']},
    {'text':'The project reproduces identically on every GPU architecture.','supported_by':[]},
]
valid_ids = {f['id'] for f in verified_findings}
for item in candidate_ai:
    refs = set(item['supported_by'])
    grounded = bool(refs) and refs <= valid_ids
    print(('GROUNDED' if grounded else 'UNSUPPORTED').ljust(12), item['text'])
```

## 3. A human-in-the-loop decision record
Grounding is necessary but not sufficient. An AI recommendation can accurately cite a finding and still be unhelpful, non-applicable, or scientifically inappropriate. Record the human decision and rationale.

```python
review_record = [
    {'finding_id':'F1','ai_priority':'high','human_decision':'accept','reason':'Environment reconstruction is required for the planned release.'},
    {'finding_id':'F2','ai_priority':'medium','human_decision':'review tests','reason':'Presence of tests does not establish scientific validation.'},
    {'finding_id':'F3','ai_priority':'high','human_decision':'accept','reason':'Published metrics need traceable provenance.'},
]
for r in review_record: print(f"{r['finding_id']} | {r['human_decision']}: {r['reason']}")
```

## 4. Responsible-use checklist
Before using AI-generated reproducibility guidance, ask: Is every factual claim traceable to evidence? Is non-applicability allowed? Can the AI change an authoritative score or only explain/prioritize findings? Are scientific judgments left to qualified humans? Are sensitive repository contents protected? Can another reviewer reconstruct why a recommendation was made?

## Hands-on exercise
Use three verified findings from one of your repositories. Ask an AI assistant—local or approved by your organization—to explain and prioritize only those findings. Require it to cite finding IDs. Label each response **supported**, **unsupported**, **overly generic**, or **requires expert review**. Do not provide confidential code or data to an unapproved service.

### Reflection
Did AI add useful explanation beyond the deterministic evidence? Did it overstate what the evidence proves? Which recommendations required domain knowledge?

AI-generated text is not reproducibility evidence. Responsible AI assistance is grounded, constrained, auditable, and subordinate to human scientific judgment. Continue to Tutorial 7, the ReproPilot case study, to examine one implementation of this design pattern.
