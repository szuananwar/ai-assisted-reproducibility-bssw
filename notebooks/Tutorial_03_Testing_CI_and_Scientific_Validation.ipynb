## 1. Three complementary layers
**Unit tests** check small software components. **Integration/workflow tests** check components working together. **Scientific validation** compares behavior with theory, trusted references, invariants, or accepted tolerances. A passing software test suite is valuable evidence but does not by itself validate a scientific conclusion.

```python
import math
def kinetic_energy(mass, velocity):
    if mass < 0:
        raise ValueError('mass must be non-negative')
    return 0.5 * mass * velocity**2
assert kinetic_energy(2.0, 3.0) == 9.0
try:
    kinetic_energy(-1.0, 2.0)
except ValueError:
    print('Unit checks passed: expected behavior and invalid input handled.')
```

## 2. Numerical validation with tolerances
Floating-point scientific results should rarely be compared with exact equality. Tolerances should come from numerical analysis, measurement uncertainty, accepted reference error, or domain knowledge—not arbitrary convenience.

```python
reference = 9.0
computed = kinetic_energy(2.0, 3.0000001)
relative_error = abs(computed-reference)/abs(reference)
tolerance = 1e-6
print(f'Computed: {computed:.9f}')
print(f'Relative error: {relative_error:.3e}')
print('Scientific tolerance check:', 'PASS' if relative_error <= tolerance else 'FAIL')
assert math.isclose(computed, reference, rel_tol=tolerance)
```

## 3. Design a CI matrix
CI can repeatedly execute tests across supported configurations. For HPC software, CI often cannot reproduce the production machine exactly, so document what is tested in CI and what still requires system-level validation.

```python
ci_plan = [
    {'check':'unit tests','frequency':'every change','purpose':'detect local regressions'},
    {'check':'workflow smoke test','frequency':'every change','purpose':'verify components integrate'},
    {'check':'reference numerical case','frequency':'every change','purpose':'detect scientific/numerical drift'},
    {'check':'large HPC validation','frequency':'release or scheduled','purpose':'validate production-scale assumptions'},
]
for row in ci_plan:
    print(f"{row['check']:25} | {row['frequency']:20} | {row['purpose']}")
```
