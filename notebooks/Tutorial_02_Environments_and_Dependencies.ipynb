## 1. Declaration versus reconstruction
A dependency file answers *what software is needed?* An environment or lock specification goes further by recording versions and resolution information. HPC workflows may additionally depend on compilers, MPI implementations, accelerator software, modules, and architecture-specific libraries.

```python
requirements = '''numpy
pandas>=2.0
scikit-learn==1.5.1
matplotlib~=3.9
'''.strip().splitlines()
def classify_requirement(req):
    if '==' in req: return 'exactly pinned'
    if '~=' in req: return 'compatible release range'
    if any(op in req for op in ['>=', '<=', '>', '<']): return 'version range'
    return 'unpinned'
for req in requirements:
    print(f"{req:25} -> {classify_requirement(req)}")
```

## 2. Build an environment record
Version pinning is not automatically the best choice for every library or project stage, but a published computational result should record enough information to reconstruct the software context used for that result.

```python
import json, platform, sys
environment_record = {
    'python': sys.version.split()[0],
    'platform': platform.platform(),
    'machine': platform.machine(),
    'dependency_specification': 'requirements.txt',
    'lock_or_resolved_environment': None,
    'compiler': None,
    'mpi': None,
    'accelerator_stack': None,
}
print(json.dumps(environment_record, indent=2))
```

## 3. Completeness check
The next cell highlights fields that still need to be documented. `None` does not always mean a defect: compiler, MPI, and accelerator fields may be non-applicable. Record non-applicability explicitly rather than silently omitting context.

```python
for key, value in environment_record.items():
    status = 'RECORDED' if value else 'REVIEW / MARK N/A'
    print(f"{key:30} {status}")
```
