## 1. What should travel with a result?
A reported metric is much more useful when it can be connected to the code revision, data/model version, preprocessing, parameters, random seed, environment, hardware, and evaluation method that produced it. Tools such as MLflow or DVC can help, but the principle is tool-independent.

```python
from datetime import datetime, timezone
import json, platform, sys
experiment = {
    'experiment_id': 'demo-001',
    'timestamp_utc': datetime.now(timezone.utc).isoformat(),
    'source_revision': 'REPLACE_WITH_GIT_COMMIT',
    'dataset': {'name':'example-dataset','version':'v1','checksum':'REPLACE_WITH_CHECKSUM'},
    'model': {'name':'example-model','version':'v1'},
    'preprocessing': ['normalize'],
    'parameters': {'learning_rate': 0.001, 'epochs': 20},
    'random_seed': 42,
    'environment': {'python':sys.version.split()[0], 'platform':platform.platform()},
    'hardware': {'device':'REPLACE_WITH_CPU_GPU_OR_HPC_RESOURCE'},
    'metrics': {'accuracy':0.91, 'weighted_f1':0.90},
}
print(json.dumps(experiment, indent=2))
```

## 2. Validate provenance completeness
The next check is intentionally simple. A real project may need domain-specific fields, licenses, access restrictions, workflow versions, compiler/MPI information, or links to archived artifacts.

```python
required = ['source_revision','dataset','parameters','random_seed','environment','hardware','metrics']
problems = []
for field in required:
    value = experiment.get(field)
    if value in (None, '', {}): problems.append(f'{field}: missing')
    elif 'REPLACE_WITH' in json.dumps(value): problems.append(f'{field}: placeholder remains')
print('Provenance review:')
print('  Complete for this tutorial.' if not problems else '\n'.join('  - '+p for p in problems))
```

## 3. Save a portable record
Uncomment the final line when working locally if you want to write the record. Avoid committing sensitive data, credentials, private dataset locations, or personally identifiable information to provenance files.

```python
provenance_json = json.dumps(experiment, indent=2)
print(provenance_json[:500] + '\n...')
# Path('provenance-demo-001.json').write_text(provenance_json, encoding='utf-8')
```
